```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Membaca dataset
data = pd.read_csv('Bank_Personal_Loan_Modelling.csv')

# Analisis distribusi kelas untuk Personal Loan
print("Distribusi Kelas Personal Loan:")
print(data['Personal Loan'].value_counts(normalize=True) * 100)

# Visualisasi bar plot
plt.figure(figsize=(6, 4))
sns.countplot(x='Personal Loan', data=data)
plt.title('Distribusi Kelas Personal Loan')
plt.xlabel('Personal Loan')
plt.ylabel('Jumlah')
plt.show()
```

```python
from sklearn.preprocessing import MinMaxScaler

# Memilih fitur numerik
numeric_features = ['Age', 'Experience', 'Income', 'CCAvg', 'Mortgage']

# Inisialisasi scaler
scaler = MinMaxScaler()

# Normalisasi fitur numerik
data[numeric_features] = scaler.fit_transform(data[numeric_features])

# Verifikasi normalisasi
print("Data setelah normalisasi fitur numerik (Personal Loan):")
print(data[numeric_features].head())
```

```python
from imblearn.over_sampling import SMOTE

# Memisahkan fitur (X) dan target (y)
X = data.drop('Personal Loan', axis=1)
y = data['Personal Loan']

# Inisialisasi SMOTE
smote = SMOTE(random_state=42)

# Terapkan SMOTE
X_balanced_pl, y_balanced_pl = smote.fit_resample(X, y)

# Perbandingan distribusi kelas
print("Distribusi Kelas Sebelum Balancing (Personal Loan):")
print(y.value_counts())
print("\nDistribusi Kelas Sesudah Balancing (Personal Loan):")
print(pd.Series(y_balanced_pl).value_counts())

# Visualisasi distribusi kelas setelah balancing
plt.figure(figsize=(6, 4))
sns.countplot(x=y_balanced_pl)
plt.title('Distribusi Kelas Personal Loan Setelah SMOTE')
plt.xlabel('Personal Loan')
plt.ylabel('Jumlah')
plt.show()
```

```python
from sklearn.model_selection import train_test_split

# Splitting data
X_train_pl, X_test_pl, y_train_pl, y_test_pl = train_test_split(
    X_balanced_pl, y_balanced_pl, test_size=0.2, stratify=y_balanced_pl, random_state=42
)

# Menampilkan ukuran dataset
print("Ukuran Training Set (Personal Loan):", X_train_pl.shape, y_train_pl.shape)
print("Ukuran Testing Set (Personal Loan):", X_test_pl.shape, y_test_pl.shape)
print("\nDistribusi Kelas di Training Set:")
print(pd.Series(y_train_pl).value_counts())
print("\nDistribusi Kelas di Testing Set:")
print(pd.Series(y_test_pl).value_counts())
```
